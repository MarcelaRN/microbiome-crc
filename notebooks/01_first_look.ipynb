```python
# load package and read in the data
import pandas as pd

pd.set_option("display.max_columns", 20)   # so wide tables don't get truncated

abundance = pd.read_csv("../data/raw/zeller2014_abundance.csv", index_col=0)
metadata = pd.read_csv("../data/raw/zeller2014_metadata.csv", index_col=0)


```

```python
# EDA (exploratory data analysis)


# Shape and structure
print("Abundance shape:", abundance.shape)   # (n_species, n_samples)
print("Metadata shape:", metadata.shape)     # (n_samples, n_variables)

# First or last rows of the file
abundance.head()      
metadata.tail(10)

#  Column names / variable names 
print(metadata.columns.tolist())

#  Data types + missing values 
metadata.info()
metadata.isnull().sum().sort_values(ascending=False)

#  Distribution of the key variable 
print(metadata["study_condition"].value_counts())   # or any other variable


#  Summary stats for numeric variables 
print(metadata.describe())

#  Do sample IDs actually match between the two files? 
set(abundance.columns) == set(metadata.index)   # True means they're aligned; False means you need to fix this before Sprint 2
```
